```python
# ===========================
# PSNR & SSIM ONLY
# ===========================

import cv2
from skimage.metrics import peak_signal_noise_ratio, structural_similarity

# ---------------------------
# 1. LOAD IMAGES
# ---------------------------
base_img_path = "C:/Users/ArselanAshraf/Desktop/Datset/01. Hazy - Raw/1400.png"
reference_img_path = "C:/Users/ArselanAshraf/Desktop/Datset/01..Hazy - Enhanced (GT)/1400.png"

base_img = cv2.imread(base_img_path)
reference_img = cv2.imread(reference_img_path)

if base_img is None:
    raise FileNotFoundError(f"Could not load: {base_img_path}")
if reference_img is None:
    raise FileNotFoundError(f"Could not load: {reference_img_path}")

# ---------------------------
# 2. CONVERT TO GRAYSCALE
# ---------------------------
base_gray = cv2.cvtColor(base_img, cv2.COLOR_BGR2GRAY)
reference_gray = cv2.cvtColor(reference_img, cv2.COLOR_BGR2GRAY)

# ---------------------------
# 3. CALCULATE PSNR & SSIM
# ---------------------------
psnr_value = peak_signal_noise_ratio(reference_gray, base_gray)
ssim_value = structural_similarity(reference_gray, base_gray)

print("PSNR:", psnr_value)
print("SSIM:", ssim_value)
```
